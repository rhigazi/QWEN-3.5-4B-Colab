```python
import os
import subprocess
import requests
import time
from huggingface_hub import hf_hub_download

os.system("pkill -f llama_cpp.server")
time.sleep(2)

MODEL_REPO = "unsloth/Qwen3.5-4B-GGUF"
MODEL_FILE = "Qwen3.5-4B-UD-Q4_K_XL.gguf"

model_path = hf_hub_download(repo_id=MODEL_REPO, filename=MODEL_FILE)

print("Starte llama_cpp.server...")
with open("server.log", "w") as log_file:
    server_process = subprocess.Popen(
        [
            "python3",
            "-m",
            "llama_cpp.server",
            "--model",
            model_path,
            "--n_gpu_layers",
            "-1",
            "--n_ctx",
            "20480",
            "--chat_format",
            "chatml",
            "--temperature",
            "0.6",
            "--top_p",
            "0.95",
            "--top_k",
            "20",
            "--port",
            "8000",
        ],
        stdout=log_file,
        stderr=log_file,
    )

print("Warte auf Server-Start...")
for i in range(40):
    try:
        res = requests.get("http://localhost:8000/v1/models")
        if res.status_code == 200:
            print("\nServer bereit!")
            break
    except requests.exceptions.ConnectionError:
        pass
    print(".", end="", flush=True)
    time.sleep(3)
```

```python
import json
import os

config_dir = os.path.expanduser("~/.pi/agent")
os.makedirs(config_dir, exist_ok=True)

config = {
    "providers": {
        "local-llama": {
            "baseUrl": "http://localhost:8000/v1",
            "apiKey": "not-needed",
            "api": "openai-completions",
            "models": [
                {
                    "id": "Qwen3.5-4B-UD-Q4_K_XL.gguf",
                    "name": "Qwen 3.5 4B",
                    "contextWindow": 20480,
                    "maxTokens": 4096,
                    "supportsTools": True,
                    # Deaktiviert das interne "Thinking" des Modells für direkte Tool-Aufrufe
                    "thinking": False,
                    # Teilt Pi mit, wie Tool-Calls geparst werden sollen
                    "toolCallType": "xml",
                    "temperature": 0.6,
                    "top_p": 0.95,
                    "top_k": 20,
                    "compat": {
                        "supportsDeveloperRole": False,
                        "supportsReasoningEffort": False,
                    },
                }
            ],
        }
    }
}

with open(os.path.join(config_dir, "models.json"), "w") as f:
    json.dump(config, f, indent=2)

print("models.json erfolgreich aktualisiert!")
```
