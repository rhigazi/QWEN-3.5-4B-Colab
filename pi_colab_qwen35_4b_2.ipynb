```python
import os
import subprocess
import time

# Ollama beenden
!pkill -f ollama

# Umgebungsvariablen setzen, um CUDA zu erzwingen
env = os.environ.copy()
env["CUDA_VISIBLE_DEVICES"] = "0"
env["OLLAMA_NUM_PARALLEL"] = "1"

# Ollama mit expliziter GPU-Umgebung starten
subprocess.Popen(["ollama", "serve"], env=env)
time.sleep(4)

# Modell im VRAM vorwärmen
!curl http://localhost:11434/api/generate -d '{"model": "qwen3.5:4b", "keep_alive": -1}'

print("Ollama wurde auf der GPU neu gestartet!")
```

```python
import json
import os

agent_dir = os.path.expanduser("~/.pi/agent")
models_path = os.path.join(agent_dir, "models.json")

models_config = {
  "providers": {
    "ollama": {
      "baseUrl": "http://localhost:11434/v1",
      "api": "openai-completions",
      "apiKey": "ollama",
      "models": [
        {
          "id": "qwen3.5:4b",
          "name": "Qwen 3.5 4B (Local)",
          "contextWindow": 60000,
          "maxTokens": 8192,
          "samplingParams": {
            "num_ctx": 60000
          }
        }
      ]
    }
  }
}

with open(models_path, "w", encoding="utf-8") as f:
    json.dump(models_config, f, indent=2)

print("Kontextfenster in models.json erfolgreich auf 60.000 Token gesetzt!")
```
