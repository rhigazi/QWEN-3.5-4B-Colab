```python
!truncate -s 0 plan.md status.md

aufgabe_text = """

### Task Description for Task Manager SPA (`taskmanager.html`)

You are tasked with creating a detailed implementation plan and a tracking file (`status.md`) for building a single-file HTML Single Page Application (`taskmanager.html`) that serves as a feature-rich Task Manager.

#### Core Technical Requirements:

1. **Zero External Dependencies:** Pure vanilla HTML, CSS, and JavaScript. No external CDNs, frameworks, or libraries.
2. **Storage:** Use browser-based internal storage (`localStorage`) so that tasks persist across page reloads.

#### Functional Requirements:

1. **Task Creation:** Create tasks with Title, Description, and a Due Date picker.
2. **Task Assignment:** Assign up to 5 different people (Person 1 to Person 5) to a task using a multi-select picklist/dropdown.
3. **Status Management:** Each task must have a status: `open`, `in_progress`, `completed`, or `archived`.
4. **Filtering & Sorting:** Filter tasks by priority and status, and sort the list by multiple criteria (e.g., due date, priority).
5. **Archiving & Deletion:** Ability to archive tasks and delete tasks permanently.

---

### Execution Instructions for the Model:

#### Step 1: Initialize Documentation (`status.md`)

* Create a file named `status.md` immediately.
* Use `status.md` as your persistent log to track every action, success, and failure/bug during implementation.
* You must always read `status.md` after a restart/new session to resume exactly where you left off.

#### Step 2: Create a Detailed Granular Plan

Break down the development of `taskmanager.html` into small, discrete, sequential steps (granular sub-tasks) that can be checked off one by one. Example breakdown:

* [ ] Task 1: Setup HTML boilerplate and CSS layout (Dashboard & Form containers).
* [ ] Task 2: Implement `localStorage` wrapper functions for CRUD operations.
* [ ] Task 3: Build Task Creation Form (Title, Description, Due Date, 5-Person Picklist).
* [ ] Task 4: Implement Task Status management (`open`, `in_progress`, `completed`, `archived`).
* [ ] Task 5: Implement Task List rendering and filtering by priority/status.
* [ ] Task 6: Implement sorting mechanisms and search/filtering options.
* [ ] Task 7: Implement task archiving and deletion features.
* [ ] Task 8: Final UI polish, error handling, and testing persistence across reloads.

#### Step 3: Execution Loop

* Execute **only one granular sub-task at a time**.
* After completing each sub-task, update `status.md` immediately with what succeeded or failed.
* Proceed to the next sub-task only after confirming the previous one works correctly."""

# In die Datei plan.md schreiben
with open("plan.md", "w", encoding="utf-8") as f:
    f.write(aufgabe_text)

print("✅ Die Aufgabe wurde erfolgreich in 'plan.md' geschrieben!")
```

```python
import subprocess

# Den Inhalt aus der plan.md einlesen
with open("plan.md", "r", encoding="utf-8") as f:
    aufgabe_text = f.read()

# Den Befehl für den pi-Agenten zusammenbauen
command = f'export PATH="/root/.pi/agent/bin:$PATH"; pi "{aufgabe_text}"'

# Ausführen und die Ausgabe direkt in execution.log schreiben
with open("execution.log", "w", encoding="utf-8") as log_file:
    process = subprocess.Popen(
        command, shell=True, stdout=log_file, stderr=subprocess.STDOUT, text=True
    )
    process.wait()

print(
    "✅ Der pi-Agent wurde ausgeführt und die Ausgabe wurde in 'execution.log' gespeichert!"
)
```
